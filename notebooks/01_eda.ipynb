Lightweight EDA notebook using `nids.data` helpers. Falls back to synthetic traffic if the full Parquet files are not present in `dataset/`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path("../src").resolve()))

from nids.config import NIDSConfig
from nids.data import generate_synthetic_traffic, load_dataset

cfg = NIDSConfig(dataset_dir=Path("../dataset"))
if cfg.train_path.exists():
    df = load_dataset(cfg.train_path)
    print(f"Loaded real dataset: {df.shape}")
else:
    df = generate_synthetic_traffic(n_samples=1000)
    print(f"Loaded synthetic demo dataset: {df.shape}")

df.head()
```

```python
print("Class distribution:")
print(df[cfg.target_column].value_counts(normalize=True).round(4))

print("\nMissing values:")
print(df.isna().sum()[df.isna().sum() > 0])
```
