Benchmarks Logistic Regression, Random Forest, HistGradientBoosting, and XGBoost (if installed) using `nids.train`.

```python
import sys
from pathlib import Path
from sklearn.model_selection import train_test_split

sys.path.insert(0, str(Path("../src").resolve()))

from nids.config import NIDSConfig
from nids.data import generate_synthetic_traffic, load_dataset
from nids.predict import predict_traffic
from nids.train import train_and_evaluate

cfg = NIDSConfig(dataset_dir=Path("../dataset"), artifacts_dir=Path("../artifacts"))

if cfg.train_path.exists() and cfg.test_path.exists():
    train_df = load_dataset(cfg.train_path)
    test_df = load_dataset(cfg.test_path)
else:
    df = generate_synthetic_traffic(n_samples=1000, random_state=cfg.random_state)
    train_df, test_df = train_test_split(df, test_size=0.25, random_state=cfg.random_state, stratify=df[cfg.target_column])

results = train_and_evaluate(train_df, test_df, config=cfg, save_artifacts=True)
print("Best model:", results["best_model_name"])
results["leaderboard"]
```

```python
sample_alerts = predict_traffic(test_df.head(10), results["best_model"], config=cfg)
sample_alerts
```
